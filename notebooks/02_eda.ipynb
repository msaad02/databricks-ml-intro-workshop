# 2. Exploratory data analysis
Mix SQL and Python in one notebook to understand the data before modeling.

**Concepts:** notebook experience, SQL + PySpark, built-in visualizations.

```python
%sql
-- Quick summary stats straight from SQL.
SELECT count(*) AS trips,
       round(avg(trip_distance), 2) AS avg_miles,
       round(avg(fare_amount), 2)   AS avg_fare
FROM samples.nyctaxi.trips
```

```python
# Load into a Spark DataFrame, then pandas for plotting.
df = spark.table("samples.nyctaxi.trips").toPandas()
df[["trip_distance", "fare_amount"]].describe()
```

```python
# Distance vs. fare — the relationship we'll model in step 3.
df.plot.scatter(x="trip_distance", y="fare_amount", alpha=0.3)
```
